```python
# write a python functions call square_sum to calculate the some of square numbers from one to any given n

n = int(input("Enter the number : "))

def square_sum(n):
  total = 0
  for i in range(1,n + 1):
    total += i ** 2
  return total

total = square_sum(n)
print(total)
```

```python
list_1 = input("Enter the number : ").split(',')
list_2 = input("Enter the number : ").split(',')

def add_corresponding(list1,list2):
  if len(list_1) != len(list_2):
    return "Error : List length is not same"
  else:
    sum_list = []
    for i in range(len(list_1)):
      sum_list.append(list_1[i] + list_2[i])
    return sum_list

print(add_corresponding(list_1,list_2))
```

```python
# Arbitary Arguments - positional arguments(*args)

def add(*args):
  total = 0
  for i in args:
    total += i
  return total

print(add(21,3,1,5))
```

```python
# Arbitary Arguments - positional arguments(*args)

def add(*args):
  for arg in args:
    print(arg)

print(add(21,3,1,5))
```

```python
def average_positive (*num):
  total = 0
  count = 0
  for i in num:
    if i > 0:
      total += i
      count += 1

  if count == 0:
    return "No positive numbers provided"
  else:
    return total / count

print(average_positive(12,2,13,54,5))
```

```python
# write a python function name average_score that,
  # (i) accept any number of keywords arguments where,
    # keys are subject names(like maths , physics)
    # values are scores obtains in each subjects
  # (ii) Calculate and print the average score of all subjects
  # (iii) If no subject are provided, then print no subject provided

def average_score(**kwargs):
  total_score = 0
  subject_count = 0
  for subject,score in kwargs.items():
    if score > 0:
      total_score += score
      subject_count += 1
    # print(key, ":", value)

  if subject_count == 0:
    return "No subject provided"
  else:
    return total_score / subject_count
    
print(average_score(maths = 90, physics = 80, chemistry = 70))
# print(average_score(maths = 50, physics = 50))
print(average_score())
```

```python
# write a function convert_to_celsius thats convert a fahrenheit temprature to celsius, rounded to two delimal places.
  #then use the map function to convert the list [32,68,77] to celsius and print the result in list

def convert_to_celsius(fahrenheit):
    celsius = (fahrenheit - 32) * 5/9
    return round(celsius, 2)

celsius_temps = [32, 68, 77]

fahrenheit_temps = list(map(convert_to_celsius, celsius_temps))

print(fahrenheit_temps)
```

```python
# Use the filter function to exact all numbers from the list [12,15,18,20,25,30,35] that are divisible by 3 but not by 5, and print the result in list.

list_1 = [12,15,18,20,25,30,35]

def divisible_by_3(x):
  return x % 3 == 0 and x % 5 != 0

result = filter(divisible_by_3,list_1)

print(list(result))
```
